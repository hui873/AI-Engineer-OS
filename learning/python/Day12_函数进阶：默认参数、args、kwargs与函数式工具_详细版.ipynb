## 2. 默认参数

常用默认行为。


```python
def calculate_fee(price, shipping_fee=10):
    return price + shipping_fee

print(calculate_fee(100))
print(calculate_fee(100, shipping_fee=0))
```

## 3. `*args`

收集任意数量位置参数。


```python
def average(*scores):
    print(type(scores))
    return sum(scores) / len(scores)

print(average(80, 90, 100))
```

## 4. `**kwargs`

收集任意数量关键字参数。


```python
def create_patient(**info):
    print(type(info))
    return info

patient = create_patient(
    name="Tom",
    age=30,
    disease="肺癌"
)

print(patient)
```

## 5. 调用时解包

list拆成位置参数，dict拆成关键字参数。


```python
def add(a, b):
    return a + b

numbers = [10, 20]
print(add(*numbers))

def show_patient(name, age):
    print(name, age)

patient = {"name": "Tom", "age": 30}
show_patient(**patient)
```

## 6. lambda、map、filter

函数式数据处理基础。


```python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x ** 2, numbers))
evens = list(filter(lambda x: x % 2 == 0, numbers))

print(squares)
print(evens)
```

# Day 8.6 - Python测试与日志系统（Testing & Logging）


## 今日总结

今天学习了 Python 工程开发中的测试和日志系统，理解了如何提高程序的可靠性，以及如何记录程序运行过程中的状态和问题。


**主要掌握：**

- 测试在工程项目中的作用
- 使用 `assert` 检查程序是否符合预期
- `assert` 与 `raise` 的区别
- 使用 `unittest` 编写自动化测试
- 测试正常情况、异常情况和边界情况
- 使用 `assertRaises()` 测试异常
- 理解测试覆盖的意义
- `logging` 模块的基本使用
- 日志等级 `DEBUG`、`INFO`、`WARNING`、`ERROR`、`CRITICAL`
- 日志格式配置
- 将日志保存到文件
- 理解模块化项目中的日志管理方式


**详细理解：**

1. **测试是保证代码可靠性的重要方式**

   理解：随着项目规模增加，手动测试每个功能会越来越困难，因此需要使用自动化测试验证代码行为，确保修改代码后不会破坏已有功能。


2. **测试不仅需要验证正确情况，也需要验证错误情况**

   理解：真实项目中用户输入可能不符合预期，因此测试需要覆盖正常数据、边界数据和异常数据，例如年龄不能为负数、文件不存在等情况。


3. **`assert`和`raise`解决的问题不同**

   理解：`assert`主要用于开发阶段检查程序内部假设，而`raise`用于正式运行过程中主动处理错误，例如数据验证失败时抛出 `ValueError`。


4. **日志系统帮助开发者理解程序运行状态**

   理解：`print()`只能临时输出信息，而 `logging` 可以记录时间、等级和具体内容，帮助开发者定位程序运行中的问题。


5. **不同日志等级代表不同问题严重程度**

   理解：`INFO`表示正常运行信息，`WARNING`表示需要关注的问题，`ERROR`表示功能失败，`CRITICAL`表示系统级严重错误，需要根据实际情况选择合适等级。


**今天最大的收获：**

> 工程代码不仅需要能够运行，还需要具备自动发现问题和记录运行状态的能力，这样才能保证项目长期维护。


**通过今天的学习，我理解了：**

- 测试不是为了证明代码永远不会出错，而是在变化中快速发现问题
- 工程项目需要提前考虑异常输入和边界情况
- 错误处理和错误记录是两个不同但互补的能力
- 日志系统是开发者了解程序运行状态的重要工具
- 从写功能代码转向构建可靠、可维护的软件系统


**今天开始从：**

> 只关注代码能否正常运行

## 2. unittest基础

完整测试类。


```python
import unittest

def calculate_bmi(weight, height):
    if height <= 0:
        raise ValueError("height must be positive")

    return weight / (height ** 2)

class TestBMI(unittest.TestCase):
    def test_normal(self):
        result = calculate_bmi(70, 1.75)

        self.assertAlmostEqual(
            result,
            22.8571428571
        )

    def test_invalid_height(self):
        with self.assertRaises(ValueError):
            calculate_bmi(70, 0)

# 在Jupyter中通常不直接调用unittest.main()
```

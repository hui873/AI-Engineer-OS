## 3. `Any` 与可空类型

现代类型写法。


```python
from typing import Any

value: Any = "Python"
patient_id: int | str = "P001"
phone: str | None = None

print(value)
print(patient_id)
print(phone)
```

## 4. dataclass

适合患者等纯数据模型。


```python
from dataclasses import dataclass

@dataclass
class Patient:
    name: str
    age: int
    disease: str | None = None

patient = Patient(
    name="Tom",
    age=30,
    disease="肺癌"
)

print(patient)
```

## 5. pathlib

路径和文件操作。


```python
from pathlib import Path

data_dir = Path("data")
data_dir.mkdir(
    parents=True,
    exist_ok=True
)

file_path = data_dir / "patients.txt"

file_path.write_text(
    "P001,Tom,30",
    encoding="utf-8"
)

print(file_path.exists())
print(file_path.read_text(encoding="utf-8"))
```

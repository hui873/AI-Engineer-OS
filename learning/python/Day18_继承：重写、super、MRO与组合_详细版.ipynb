## 1. 基础继承

保留原ElectricCar代码。


```python
#继承
#继承是面向对象编程中的一个重要概念，它允许一个类（子类）继承另一个类（父类）的属性和方法，从而实现代码的重用和扩展。
# 1，父类（基类）：被继承的类称为父类
# 2，子类（派生类）：继承父类的类称为子类
# 3，继承的语法：class 子类名(父类名):
#汽车类
class Car:
    def __init__(self, c_color, c_brand, c_model):
        self.color = c_color
        self.brand = c_brand
        self.model = c_model

    def start(self):  # 启动
        print(f'{self.brand}{self.model}正在启动...')
    def stop(self):  # 停止
        print(f'{self.brand}{self.model}停止行驶...')
    def run(self):  # 行驶
        print(f'{self.brand}{self.model}正在行驶')

# 电动车类继承汽车类
class ElectricCar(Car):  
    def __init__(self, c_color, c_brand, c_model, battery_capacity):
        super().__init__(c_color, c_brand, c_model)  # 调用父类的初始化方法
        self.battery_capacity = battery_capacity  # 电池容量

    def charge(self):  # 充电
        print(f'{self.brand}{self.model}正在充电，电池容量为{self.battery_capacity}kWh')

if __name__ == "__main__":
    my_electric_car = ElectricCar("红色", "特斯拉", "Model 3", 75)
    my_electric_car.start()
    my_electric_car.run()
    my_electric_car.charge()
    my_electric_car.stop()
```

## 2. 方法重写与父类调用

保留原代码。


```python
#继承，子类继承父类的属性和方法，重写父类的方法，添加新的方法，多继承，super()函数
class Car:
    def __init__(self, c_color, c_brand, c_model):
        self.color = c_color
        self.brand = c_brand
        self.model = c_model

    def start(self):  # 启动
        print(f'{self.brand}{self.model}正在启动...')
    def stop(self):  # 停止
        print(f'{self.brand}{self.model}停止行驶...')
    def run(self):  # 行驶
        print(f'{self.brand}{self.model}正在行驶')
    def charge(self):  # 充电
        print(f'{self.brand}{self.model}正在加燃料')

# 电动车类继承汽车类
class ElectricCar(Car):  
    def charge(self):
        super().charge()  # 调用父类的charge方法
        print(f'{self.brand}{self.model}正在充电')

class GasCar(Car):
    def charge(self):
        Car.charge(self)  # 调用父类的charge方法
        print(f'{self.brand}{self.model}正在加汽油')

if __name__ == "__main__":
    my_electric_car = ElectricCar("红色", "特斯拉", "Model 3",)
    my_electric_car.charge()

    my_gas_car = GasCar("蓝色", "宝马", "X5")

    my_gas_car.charge()
    
    
```

## 3. 多继承与MRO

保留原多继承代码。


```python
#多继承
class Car:
    def __init__(self, c_color, c_brand, c_model)->None:
        self.color = c_color
        self.brand = c_brand
        self.model = c_model

    def start(self):  # 启动
        print(f'{self.brand}{self.model}正在启动...')
    def stop(self):  # 停止
        print(f'{self.brand}{self.model}停止行驶...')
    def run(self):  # 行驶
        print(f'{self.brand}{self.model}正在行驶')
    def charge(self):  # 充电
        print(f'{self.brand}{self.model}正在加燃料')

class HuaweiAiDriving:#2用法
    """华为AI智能驾驶"""
    def __init__(self,version="V1.0"):
        self.version=version
    def run(self):
        print(f'使用华为Ai智能驾驶系统{self.version}正在行驶...')

# 电动车类继承汽车类
class ElectricCar(Car, HuaweiAiDriving):  #MRO（method resolution order，方法解析顺序）
    #-->从左到右，先找Car类，再找HuaweiAiDriving类
    """电动车类"""
    def __init__(self, c_color, c_brand, c_model,version="V2.0"):
        super().__init__(c_color, c_brand, c_model)  # 调用父类的初始化方法
        HuaweiAiDriving.__init__(self,version)  # 调用父类的初始化方法
        

    def charge(self):
        super().charge()  # 调用父类的charge方法
        print(f'{self.brand}{self.model}正在充电')
    def run(self):
        super().run()  # 调用父类的run方法
        print(f'使用华为Ai智能驾驶系统{self.version}正在行驶...')
class GasCar(Car):
    def charge(self):
        Car.charge(self)  # 调用父类的charge方法
        print(f'{self.brand}{self.model}正在加汽油')

if __name__ == "__main__":
    my_electric_car = ElectricCar("红色", "问界", "Model 3",)
    my_electric_car.run()


              
```

## 4. 查看MRO

观察方法解析顺序。


```python
class A:
    pass

class B:
    pass

class C(A, B):
    pass

print(C.mro())
print(C.__mro__)
```

## 5. 组合

has-a关系示例。


```python
class AiDrivingSystem:
    def run(self):
        print("AI driving system running")

class Car:
    def __init__(self):
        self.ai_system = AiDrivingSystem()

    def run_ai(self):
        self.ai_system.run()

car = Car()
car.run_ai()
```

# 学习内容与代码实践


## 1. 字符串基础与遍历

字符串是有序字符序列。


```python
#介绍：字符串是字符的容器，一个字符串中可以存放任意数量的字符，每个字符都有一个索引位置
#字符串不可修改，只能重新赋值
str1 = "hello world"
print(*str1)
for i in str1:
    print(i,end=' ')
```

## 2. 字符串常用方法

查找、统计、大小写、分割、清理和替换。


```python
#字符串常用方法
str = "hello-world"

# find()在字符串中查找子串，返回第一次出现的索引位置，找不到返回-1
print(str.find("world"))
# count()统计子串在字符串中出现的次数
print(str.count("l"))
# upper()将字符串中的所有字母转换为大写
print(str.upper())
# lower()将字符串中的所有字母转换为小写
print(str.lower())
# split()将字符串按指定分隔符分割成列表
print(str.split('-'))
# strip()去除字符串两端的空白字符或指定字符
print(str.strip())
# replace()将字符串中的指定子串替换为新的子串
print(str.replace("world","python"))
# startswith()检查字符串是否以指定子串开头
print(str.startswith("hello"))
```

## 3. 邮箱基础验证

通过字符串方法建立简单规则。


```python
# 邮箱格式验证：用户输入一个邮箱，验证邮箱格式是否正确（包含一个@和至少一个.），如果输入正确，
# 输出"邮箱格式正确”，否则输出”邮箱格式错误"。

# 获取用户输入的邮箱字符串
shuru=input("请输入邮箱：")

# 统计输入中 '@' 和 '.' 的数量以验证基本格式
num1=shuru.count("@")
num2=shuru.count(".")

# 判断是否恰好包含一个 '@' 且至少包含一个 '.'
if num1==1 and num2>=1:
    print(f'{shuru}邮箱格式正确')
else:
   print(f'{shuru}邮箱格式错误')
```

## 4. 回文判断

利用反向切片 `[::-1]`。


```python
# 输入一个字符串，判断该字符串是否是回文（两边对称）
# 获取用户输入的字符串
s = input("请输入一个字符串: ")

# 使用切片 [::-1] 将字符串反转，然后与原字符串比较
# 如果相等，说明是回文
if s == s[::-1]:
    print(f"'{s}' 是回文")
else:
    print(f"'{s}' 不是回文")

# 黄山落叶松叶落山黄
# 上海自来水来自海上




```

## 5. 批量字符串处理

字符串、列表、循环与 `enumerate()` 组合。


```python
#将用户输入的10个字符串，反转后全部转换为大写，然后记录在列表中，最后将列表内容，遍历输出出来。

# 创建空列表用于存储处理后的字符串
result_list = []

# 循环获取10个字符串输入
for i in range(10):
    # 获取用户输入
    s = input(f"请输入第{i+1}个字符串: ")
    # 反转字符串并转换为大写
    processed = s[::-1].upper()
    # 添加到结果列表
    result_list.append(processed)

# 遍历输出列表内容
print("处理后的字符串列表:")
for i, s in enumerate(result_list, 1):
    print(f"{i}. {s}")
```

## 6. `split()` 与 `join()`

这是文本拆分与重组的核心组合。


```python
symptoms_text = "fever,cough,headache"

# split()：字符串 -> list
symptoms = symptoms_text.split(",")
print(symptoms)

# join()：多个字符串 -> 一个字符串
joined = " | ".join(symptoms)
print(joined)
```

## 7. `startswith()` 与 `endswith()`

用于判断文本是否具有某个固定开头或结尾。


```python
filename = "patients.csv"

print(filename.startswith("patient"))
print(filename.endswith(".csv"))

url = "https://example.com"
print(url.startswith("https://"))
```

## 8. 字符串不可变性

字符串方法不会原地修改原字符串。


```python
text = "  Hello World  "

# 这些方法都会返回新的str
clean_text = text.strip()
lower_text = clean_text.lower()
new_text = lower_text.replace("world", "python")

print(text)
print(clean_text)
print(lower_text)
print(new_text)
```

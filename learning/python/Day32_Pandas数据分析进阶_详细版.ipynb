## 3. 缺失值检查

先识别问题。


```python
import pandas as pd

df = pd.DataFrame({
    "type": ["A", None, "B"],
    "height": [1.7, 1.8, None]
})

print(df.isna())
print(df.isna().sum())
```

## 4. `groupby()` 与 `value_counts()`

分组统计和频数统计。


```python
import pandas as pd

df = pd.DataFrame({
    "department": ["A", "A", "B", "B"],
    "age": [20, 30, 40, 50]
})

print(
    df.groupby("department")["age"].mean()
)

print(
    df["department"].value_counts()
)
```

## 5. `merge()`

连接患者表与诊断表。


```python
import pandas as pd

patients = pd.DataFrame({
    "patient_id": ["P001", "P002"],
    "name": ["Tom", "Alice"]
})

visits = pd.DataFrame({
    "patient_id": ["P001", "P002"],
    "diagnosis": ["A", "B"]
})

merged = patients.merge(
    visits,
    on="patient_id",
    how="left"
)

print(merged)
```

## 6. `map()` 与 `apply()`

补齐自定义映射。


```python
import pandas as pd

df = pd.DataFrame({
    "age": [17, 30, 70]
})

df["group"] = df["age"].map(
    lambda age: "adult"
    if 18 <= age < 65
    else "other"
)

df["age_next_year"] = df["age"].apply(
    lambda age: age + 1
)

print(df)
```

# 后续按需补充说明

以下内容**不放入当前主线学习**，因为现在进入医疗系统项目并不需要先掌握，但以后遇到真实需求时值得回来学习。

## 1. Debug 调试

建议在项目出现真正复杂Bug后系统学习：

* Breakpoint
* Step Over
* Step Into
* Step Out
* Variables
* Watch
* Call Stack
* Exception Breakpoint

学习目标不是记按钮，而是建立“复现问题 → 下断点 → 观察状态 → 缩小范围 → 验证修复”的调试流程。

## 2. `asyncio`

当以后遇到大量网络I/O、并发API请求时再学习：

* `async def`
* `await`
* `asyncio.run()`
* coroutine
* Task
* 并发等待I/O

当前不需要研究event loop底层。

## 3. `requests`

进入HTTP API、LLM API、RAG或Agent阶段时学习：

* GET / POST
* query parameters
* headers
* JSON body
* `response.json()`
* timeout
* HTTP status code
* 网络异常处理

## 4. Pydantic

进入FastAPI、LLM structured output、Agent工具参数校验时学习：

* `BaseModel`
* 字段类型
* 默认值
* 可选字段
* ValidationError
* 模型嵌套
* 序列化

## 5. `copy.deepcopy()`

如果项目出现复杂嵌套list/dict/对象复制问题，再重点学习：

* 引用
* 浅拷贝
* 深拷贝
* NumPy view/copy

## 6. 更深入的pytest

项目规模变大后再学习：

* fixture
* mock
* monkeypatch
* 测试隔离
* 集成测试

## 7. `pyproject.toml`

当前认识即可。以后项目工具增多后再学习如何统一管理：

* 项目元数据
* 依赖
* Black
* Ruff
* pytest
* build配置

## 8. 多线程与多进程

当前不作为Python基础必修。

以后遇到CPU并行、后台任务或I/O并发需求时再学习。

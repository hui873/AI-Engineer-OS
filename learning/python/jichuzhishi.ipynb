```python
'''if的高阶使用
#输入三角形的三边，判断是等边三角形，还是直角三角形还是普通三角形，或者不能构成三角形
# 获取用户输入的三个整数，分别代表三角形的三条边长
'''
a,b,c=map(int,input("请输入三角形的三边：").split())
# 判断三条边是否满足三角形不等式定理（任意两边之和大于第三边）
if a+b>c and a+c>b and b+c>a:
    # 判断是否为等边三角形（三边相等）
    if a==b and b==c:
        print("等边三角形")
    # 判断是否为直角三角形（满足勾股定理）
    elif a**2+b**2==c**2 or a**2+c**2==b**2 or b**2+c**2==a**2:
        print("直角三角形")
    # 既非等边也非直角的普通三角形
    else:
        print("普通三角形")
# 不满足三角形构成条件
else:
    print("不能构成三角形")
```

```python
#将用户输入的10个字符串，反转后全部转换为大写，然后记录在列表中，最后将列表内容，遍历输出出来。

# 创建空列表用于存储处理后的字符串
result_list = []

# 循环获取10个字符串输入
for i in range(10):
    # 获取用户输入
    s = input(f"请输入第{i+1}个字符串: ")
    # 反转字符串并转换为大写
    processed = s[::-1].upper()
    # 添加到结果列表
    result_list.append(processed)

# 遍历输出列表内容
print("处理后的字符串列表:")
for i, s in enumerate(result_list, 1):
    print(f"{i}. {s}")
```

```python
# 根据提供的学生成绩单，完成如下需求：
# 1，计算每个学生的总分、各科平均分，然后一并输出出来。
# 2．统计各科成绩的最低分、最高分、平均分，并输出。
# 3．查找成绩优秀（平均分大于90）的学生，并输出。
# 定义学生数据：(学号, 姓名, 语文, 数学, 英语)
students = [
    ("S001", "王伟", 85, 92, 78),
    ("S002", "王林", 85, 92, 78),
    ("S003", "李慕婉", 92, 88, 95),
    ("S004", "十三", 78, 85, 82),
    ("S005", "曾牛", 88, 79, 91),
    ("S006", "周轶", 95, 96, 89),
    ("S007", "王卓", 76, 82, 77),
    ("S008", "红蝶", 89, 91, 94),
    ("S009", "徐立国", 75, 69, 82),
    ("S010", "许木", 86, 89, 98),
    ("S011", "遁天", 66, 59, 72)
]

print("=== 需求1：每个学生的总分和平均分 ===")
for student in students:
    student_id, name, chinese, math, english = student
    total = chinese + math + english
    average = total / 3
    print(f"{name}({student_id}): 总分={total}, 平均分={average:.2f}")

print("\n=== 需求2：各科成绩统计 ===")
chinese_scores = [s[2] for s in students]
math_scores = [s[3] for s in students]
english_scores = [s[4] for s in students]

print(f"语文最低分：{min(chinese_scores)}，最高分：:{max(chinese_scores)},平均分:{sum(chinese_scores)/len(chinese_scores):.2f}")
print(f"数学最低分:{min(math_scores)}，最高分:{max(math_scores)}，平均分：{sum(math_scores)/len(math_scores):.2f}") 
print(f"英语最低分:{min(english_scores)}，最高分：:{max(english_scores)},平均分:{sum(english_scores)/len(english_scores):.2f}")

print("\n=== 需求3：成绩优秀的学生（平均分>90）===")
excellent_students = []
for student in students:
    student_id, name, chinese, math, english = student
    total = chinese + math + english
    average = total / 3
    if average > 90:
        excellent_students.append((name, student_id, average))

if excellent_students:
    for name, student_id, avg in excellent_students:
        print(f"{name}({student_id}): 平均分={avg:.2f}")
else:
    print("没有平均分大于90的学生")
```

```python
# 1，定义一个函数：根据传入的底和高计算三角形面积的函数（三角形面积=底*高/2）。
# 2．定义一个函数：计算传入的字符串中元音字母的个数（元音字母为aeiouAEIou）。
# 3。定义一个函数：计算传入的班级学员高考成绩列表中成绩的最高分、最低分、平均分（保留1位小数），
# 并返回。
def triangle_area(base,height):
    """
    :param base: 三角形的底
    :param height: 三角形的高
    :return: 三角形的面积
    """
    return base*height/2

def count_vowels(s):
    """
    :param s: 输入的字符串
    :return: 元音字母的个数
    """
    vowels = "aeiouAEIOU"
    count = 0
    for char in s:
        if char in vowels:
            count += 1
    return count    
def calculate_scores(scores):
    """
    :param scores: 输入的班级学员高考成绩列表
    :return: 最高分、最低分、平均分（保留1位小数）
    """
    if not scores:
        return None
    highest = max(scores)
    lowest = min(scores)
    average = round(sum(scores)/len(scores),1)
    return highest,lowest,average
```

```python
# 定义一个函数，根据传入的数字，计算该数字阶乘的结果
def factorial(n):
    """
    :param n: 输入的数字
    :return: 该数字的阶乘
    """
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
print(factorial(5))  # 输出: 120
```

```python
# 定义一个函数，用于根据传入的一批商品信息（商品名、价格、数量）、优惠（优惠券、积分抵扣）、运费信息计算订单的总金额。
# 具体规则如下：
# ·优惠券需要商品金额满5000才可以使用，且优惠券金额不能超过商品总价。
# ·积分抵扣需要商品总金额满5000才可以使用，100积分抵扣1元（且抵扣金额不能超过商品总价，积分只能整百抵扣）。
def calculate_order_total(items, coupon, points, shipping_fee):
    """
    :param items: 商品信息列表，每个元素为一个字典，包含商品名、价格、数量
    :param coupon: 优惠券金额
    :param points: 积分抵扣金额（100积分抵扣1元）
    :param shipping_fee: 运费金额
    :return: 订单总金额
    """
    total_price = sum(item['price'] * item['quantity'] for item in items)
    
    # 检查优惠券使用条件
    if total_price >= 5000 and coupon <= total_price:
        total_price = total_price - coupon
    
    # 检查积分抵扣使用条件
    if total_price >= 5000 and points >= 100:
        points_deduction = (points // 100)  # 每100积分抵扣1元
        if points_deduction > total_price:
            points_deduction = total_price
        total_price -= points_deduction
    
    # 加上运费
    total_price += shipping_fee
    
    return total_price
print(calculate_order_total(items=[{'name': '商品A', 'price': 2000, 'quantity': 5}, {'name': '商品B', 'price': 1000, 'quantity': 3}],
                             coupon=500, points=1030, shipping_fee=100))
print(calculate_order_total(items=[{'name': '商品A', 'price': 2000, 'quantity': 1}, {'name': '商品B', 'price': 1000, 'quantity': 1}], 
                            coupon=500, points=1030, shipping_fee=19.9))
```

```python
'''if的高阶使用
#输入三角形的三边，判断是等边三角形，还是直角三角形还是普通三角形，或者不能构成三角形
# 获取用户输入的三个整数，分别代表三角形的三条边长
'''
a,b,c=map(int,input("请输入三角形的三边：").split())
# 判断三条边是否满足三角形不等式定理（任意两边之和大于第三边）
if a+b>c and a+c>b and b+c>a:
    # 判断是否为等边三角形（三边相等）
    if a==b and b==c:
        print("等边三角形")
    # 判断是否为直角三角形（满足勾股定理）
    elif a**2+b**2==c**2 or a**2+c**2==b**2 or b**2+c**2==a**2:
        print("直角三角形")
    # 既非等边也非直角的普通三角形
    else:
        print("普通三角形")
# 不满足三角形构成条件
else:
    print("不能构成三角形")
```

```python
# 根据提供的学生成绩单，完成如下需求：
# 1，计算每个学生的总分、各科平均分，然后一并输出出来。
# 2．统计各科成绩的最低分、最高分、平均分，并输出。
# 3．查找成绩优秀（平均分大于90）的学生，并输出。
# 定义学生数据：(学号, 姓名, 语文, 数学, 英语)
students = [
    ("S001", "王伟", 85, 92, 78),
    ("S002", "王林", 85, 92, 78),
    ("S003", "李慕婉", 92, 88, 95),
    ("S004", "十三", 78, 85, 82),
    ("S005", "曾牛", 88, 79, 91),
    ("S006", "周轶", 95, 96, 89),
    ("S007", "王卓", 76, 82, 77),
    ("S008", "红蝶", 89, 91, 94),
    ("S009", "徐立国", 75, 69, 82),
    ("S010", "许木", 86, 89, 98),
    ("S011", "遁天", 66, 59, 72)
]

print("=== 需求1：每个学生的总分和平均分 ===")
for student in students:
    student_id, name, chinese, math, english = student
    total = chinese + math + english
    average = total / 3
    print(f"{name}({student_id}): 总分={total}, 平均分={average:.2f}")

print("\n=== 需求2：各科成绩统计 ===")
chinese_scores = [s[2] for s in students]
math_scores = [s[3] for s in students]
english_scores = [s[4] for s in students]

print(f"语文最低分：{min(chinese_scores)}，最高分：:{max(chinese_scores)},平均分:{sum(chinese_scores)/len(chinese_scores):.2f}")
print(f"数学最低分:{min(math_scores)}，最高分:{max(math_scores)}，平均分：{sum(math_scores)/len(math_scores):.2f}") 
print(f"英语最低分:{min(english_scores)}，最高分：:{max(english_scores)},平均分:{sum(english_scores)/len(english_scores):.2f}")

print("\n=== 需求3：成绩优秀的学生（平均分>90）===")
excellent_students = []
for student in students:
    student_id, name, chinese, math, english = student
    total = chinese + math + english
    average = total / 3
    if average > 90:
        excellent_students.append((name, student_id, average))

if excellent_students:
    for name, student_id, avg in excellent_students:
        print(f"{name}({student_id}): 平均分={avg:.2f}")
else:
    print("没有平均分大于90的学生")
```

```python
# 1，定义一个函数：根据传入的底和高计算三角形面积的函数（三角形面积=底*高/2）。
# 2．定义一个函数：计算传入的字符串中元音字母的个数（元音字母为aeiouAEIou）。
# 3。定义一个函数：计算传入的班级学员高考成绩列表中成绩的最高分、最低分、平均分（保留1位小数），
# 并返回。
def triangle_area(base,height):
    """
    :param base: 三角形的底
    :param height: 三角形的高
    :return: 三角形的面积
    """
    return base*height/2

def count_vowels(s):
    """
    :param s: 输入的字符串
    :return: 元音字母的个数
    """
    vowels = "aeiouAEIOU"
    count = 0
    for char in s:
        if char in vowels:
            count += 1
    return count    
def calculate_scores(scores):
    """
    :param scores: 输入的班级学员高考成绩列表
    :return: 最高分、最低分、平均分（保留1位小数）
    """
    if not scores:
        return None
    highest = max(scores)
    lowest = min(scores)
    average = round(sum(scores)/len(scores),1)
    return highest,lowest,average
```

```python
# 定义一个函数，根据传入的数字，计算该数字阶乘的结果
def factorial(n):
    """
    :param n: 输入的数字
    :return: 该数字的阶乘
    """
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
print(factorial(5))  # 输出: 120
```

```python
# 定义一个函数，用于根据传入的一批商品信息（商品名、价格、数量）、优惠（优惠券、积分抵扣）、运费信息计算订单的总金额。
# 具体规则如下：
# ·优惠券需要商品金额满5000才可以使用，且优惠券金额不能超过商品总价。
# ·积分抵扣需要商品总金额满5000才可以使用，100积分抵扣1元（且抵扣金额不能超过商品总价，积分只能整百抵扣）。
def calculate_order_total(items, coupon, points, shipping_fee):
    """
    :param items: 商品信息列表，每个元素为一个字典，包含商品名、价格、数量
    :param coupon: 优惠券金额
    :param points: 积分抵扣金额（100积分抵扣1元）
    :param shipping_fee: 运费金额
    :return: 订单总金额
    """
    total_price = sum(item['price'] * item['quantity'] for item in items)
    
    # 检查优惠券使用条件
    if total_price >= 5000 and coupon <= total_price:
        total_price = total_price - coupon
    
    # 检查积分抵扣使用条件
    if total_price >= 5000 and points >= 100:
        points_deduction = (points // 100)  # 每100积分抵扣1元
        if points_deduction > total_price:
            points_deduction = total_price
        total_price -= points_deduction
    
    # 加上运费
    total_price += shipping_fee
    
    return total_price
print(calculate_order_total(items=[{'name': '商品A', 'price': 2000, 'quantity': 5}, {'name': '商品B', 'price': 1000, 'quantity': 3}],
                             coupon=500, points=1030, shipping_fee=100))
print(calculate_order_total(items=[{'name': '商品A', 'price': 2000, 'quantity': 1}, {'name': '商品B', 'price': 1000, 'quantity': 1}], 
                            coupon=500, points=1030, shipping_fee=19.9))
```

```python
# 异常（也称为Bug）就是程序运行过程中出现的错误，它会中断程序的正常执行流程。
# 作用：保证数据、逻辑的正确性，避免程序执行混乱
# 在开发阶段，尽量发现更多的问题，尽早解决问题，保障程序正常执行

try:
    a = 10 / 0
    print(a)
except ZeroDivisionError as h:# 捕获除数为0的异常，打印异常信息h
    print("除数不能为0",h)
except Exception as e:# 捕获其他异常，打印异常信息e
    print("发生异常:", e)
finally:
    print("程序执行结束")
# 程序运行过程中出现异常，有两种处理方案：
# 1．不做处理：整个程序因为一个Bug，中断执行。
# 2，捕获异常：按照我们自己的处理方式，处理完异常，程序继续执行。
```

```python
# 异常传递就是异常在函数调用中层层上报的过程，直到有人处理它，或者程序崩溃。
def fun1():
    print("fun1.. running.")
    fun2()

def fun2():
    print("fun2 .. running...")
    fun3()

def fun3():
    print("fun3... running")
    # 这里会抛出异常
    print(my.color)
    raise Exception("fun3 函数抛出异常")
if __name__ == "__main__":
    try:
        fun1()
    except Exception as e:
        print("程序运行出错了，请联系管理员，错误信息：", e)
```

```python
#自定义异常的例子
class MyError(Exception):
    """自定义异常类"""

    def __init__(self, message, code=1001):
        """
        初始化自定义异常对象。

        :param message: 异常描述信息
        :param code: 异常错误码，默认为 1001
        """
        self.message = message  # 保存异常说明
        self.code = code        # 保存异常错误码
        super().__init__(message)  # 调用父类 Exception 的初始化方法


def check_age(age):
    """
    检查年龄是否合法。

    :param age: 输入的年龄
    :return: 如果年龄合法，返回提示信息
    :raises MyError: 当年龄不合法时抛出自定义异常
    """
    if age < 0:
        # 如果年龄小于 0，说明输入不合法
        raise MyError("年龄不能为负数", code=1001)

    if age < 18:
        # 如果年龄小于 18，说明用户未成年
        raise MyError("未成年人不能使用", code=1002)

    # 以上条件都不满足，说明年龄合法
    return "允许使用"


try:
    # 尝试调用函数，并传入一个非法年龄
    print(check_age(-3))
except MyError as e:
    # 捕获自定义异常，并输出异常信息
    print(f"捕获到异常：{e.message}")
    print(f"错误码：{e.code}")
```

## 今日总结

* 使用 `indent=4` 格式化 JSON 文件
* 使用 `ensure_ascii=False` 正常显示中文

**详细理解：**

1. **JSON作为数据交换和存储格式**

   理解：JSON 可以把程序中的字典、列表等数据保存到外部文件，也可以在不同程序之间传递数据，是程序内部数据与外部存储之间的重要桥梁。

2. **Python数据与JSON数据需要进行转换**

   理解：Python 中的字典、列表、字符串、数字、布尔值和 `None` 可以转换为 JSON 支持的数据类型，但两者属于不同的数据表示形式，需要通过序列化和反序列化完成转换。

```python
#csv文件操作
# 已知 patients.csv 内容如下：
# name,age,gender,blood_sugar
# 张三,17,男,6.2
# 李四,25,女,7.5
# 王五,30,男,8.1
# 赵六,22,女,5.8
# 孙七,40,女,7.2
# 请独立完成一个 Python 程序，实现以下流程：
# 使用 csv.DictReader() 读取 patients.csv
# 将年龄转换为整数，将血糖值转换为浮点数
# 筛选同时满足以下条件的患者：
# 年龄大于或等于 18 岁
# 血糖值大于或等于 7.0
# 为符合条件的患者增加字段：row["status"] = "需复查"
# 将筛选结果保存到 review_patients.csv
# 新文件字段顺序为：["name", "age", "gender", "blood_sugar", "status"]

import csv

review_patients = []
with open("patients.csv",mode="r",encoding="utf-8") as infile:
    reader = csv.DictReader(infile)# 以字典的方式读取内容

    for row in reader:
        row["age"] = int(row["age"])
        row["blood_sugar"] = float(row["blood_sugar"])

        if row["age"] >= 18 and row["blood_sugar"] >= 7.0:
            row["status"] = "需复查"
            review_patients.append(row)

with open("review_patients.csv",mode="w",newline="",encoding="utf-8") as outfile:
    # 写入文件头
    fieldnames = [
        "name",
        "age",
        "gender",
        "blood_sugar",
        "status"
    ]

    writer = csv.DictWriter(outfile, fieldnames=fieldnames)# 以字典的方式写入内容
    writer.writeheader()
    writer.writerows(review_patients)
```

**主要掌握：**

* CSV 与 JSON 适用场景的区别

* 使用 `csv.writer()` 写入列表数据
* 使用 `csv.DictWriter()` 写入字典数据
* 使用 `writeheader()`、`writerow()` 和 `writerows()` 写入数据
* 使用 `"w"` 和 `"a"` 模式覆盖或追加数据
* 使用 `newline=""` 避免写入多余空行
* 使用 `encoding="utf-8-sig"` 提高 Excel 对中文 CSV 的兼容性
* 对 CSV 中的字符串进行 `int()` 和 `float()` 类型转换
* 完成数据筛选、字段修改、结果收集与统一保存

**详细理解：**

1. **CSV适合保存结构固定的二维表格数据**

   理解：当每条数据拥有相同字段，并且可以用行和列表示时，CSV 比较直观，也便于使用 Excel、Pandas 和机器学习工具处理；存在多层嵌套关系的数据则更适合使用 JSON。

2. **`csv.reader()`与`csv.DictReader()`提供不同的数据访问方式**

   理解：`csv.reader()` 将每一行读取为列表，需要通过数字下标访问字段；`csv.DictReader()` 将表头作为字典的键，可以通过字段名称访问数据，因此在字段较多的项目中更加清晰和易于维护。

3. **CSV读取后的内容默认都是字符串**

   理解：CSV 本质上是文本文件，不会保存整数、浮点数等类型信息，因此进行年龄判断或血糖值比较前，需要使用 `int()` 或 `float()` 完成类型转换。

4. **数据转换不等于修改原始字典**

   理解：执行 `age = int(row["age"])` 只是创建了一个转换后的新变量，并没有改变 `row["age"]`；只有显式执行 `row["age"] = int(row["age"])`，才会修改字典中的原数据。

5. **文件写入应该与数据处理职责分开**

   理解：在循环中反复打开文件并调用 `writeheader()` 会造成重复表头和不必要的资源操作，更合理的流程是先读取、转换和筛选数据，再将结果统一写入输出文件。

# 3Blue1Brown 线性代数本质 Day 3 学习总结（第5～7集）

## 第五集：三维空间中的线性变换

**主要掌握：**

* 线性变换可以从二维空间扩展到三维以及更高维空间
* 矩阵可以表示空间中的旋转、缩放、拉伸和剪切等变化

* 一个线性变换可以通过观察基向量的变化来理解整体变化
* 高维数据处理的本质也是对空间进行变换


**详细理解：**

1. **线性变换的本质**

   理解：线性变换不是简单地改变向量中的数字，而是在保持线性关系的情况下改变整个空间。一个空间中的所有点都会按照同一种规则发生变化。


2. **矩阵与空间变化的关系**

   理解：矩阵的作用可以看成对基向量进行变换。只要知道基向量经过变化后的位置，就可以确定整个空间如何变化。


3. **高维空间中的矩阵**

   理解：虽然人类难以直接观察高维空间，但数学上仍然可以通过矩阵描述高维数据的变化过程。这也是机器学习中处理高维特征的基础。

* 深度学习中的特征提取本质上也是不断进行空间转换

## 第六集：行列式（Determinant）

**主要掌握：**

* 行列式表示线性变换对空间大小的影响
* 行列式可以理解为面积或体积变化的比例
* 行列式为正表示空间方向没有翻转
* 行列式为负表示空间发生方向翻转
* 行列式为 0 表示空间被压缩，信息发生丢失


**详细理解：**

1. **行列式与空间变化**

   理解：矩阵进行变换时，不只是改变单个向量，而是改变整个空间。行列式描述的是这种变化对空间面积或体积造成的影响。


2. **行列式为 0 的意义**

   理解：当一个二维空间被压缩成一条线，或者三维空间被压缩成一个平面时，说明部分维度的信息消失，因此无法恢复。


3. **行列式与信息保留**

   理解：行列式可以帮助判断一个变换是否损失空间信息。在机器学习中，特征经过转换后是否保留有效信息也是类似的问题。


---

## 第七集：逆矩阵（Inverse Matrix）

**主要掌握：**

* 逆矩阵表示一个线性变换的反向过程
* 矩阵与逆矩阵结合可以恢复原始状态
* 不是所有矩阵都有逆矩阵
* 信息丢失的变换无法通过逆矩阵恢复


**详细理解：**

1. **逆矩阵的意义**

   理解：如果一个矩阵表示一次空间变化，那么逆矩阵就是执行相反变化的工具。两次变化结合后，可以回到原来的空间。


2. **为什么有些矩阵不可逆**

   理解：如果矩阵把二维空间压缩成一维空间，那么不同输入可能得到相同结果，导致原始信息无法确定，因此不存在恢复方法。


3. **可逆变换与信息保存**

   理解：矩阵是否可逆，本质上反映了数据经过转换后是否仍然保留完整信息。


**今天最大的收获：**

> 一个矩阵变换不仅改变数据形式，更重要的是决定信息是否被保留以及是否能够恢复。


**通过今天的学习，我理解了：**

* 行列式描述空间变化带来的信息影响
* 逆矩阵代表恢复信息的能力
* 数据转换过程中最重要的问题之一是信息是否丢失

# 3Blue1Brown 线性代数本质 Day 5 学习总结（第8～10集）

   理解：矩阵接收输入后能够产生的所有结果组成列空间，它代表这个矩阵可以表达的信息范围。


2. **零空间代表丢失的信息**

   理解：如果某些输入经过矩阵变换后变成零，说明这些方向的信息被矩阵完全消除，这些方向组成零空间。


3. **空间与信息的关系**

   理解：矩阵变换不仅是在改变数据形式，也是在选择哪些信息被保留，哪些信息被忽略。

**详细理解：**

1. **条件判断负责选择执行路径**

   理解：程序不会永远按同一条路线运行，而是根据布尔条件决定哪些代码执行。

2. **复杂业务规则可以拆成多个条件**

   理解：闰年和三角形判断本质上都是将业务规则翻译成比较与逻辑表达式。

3. **循环负责重复执行任务**

   理解：`for` 更适合遍历一组确定的数据，`while` 更适合根据某个状态决定是否继续。

4. **循环必须考虑结束条件**

   理解：尤其是 `while`，如果影响条件的变量永远不变化，就可能形成无限循环。

5. **嵌套循环表达二维结构**

   理解：外层循环控制一层结构，内层循环控制另一层结构，这种思维以后会直接进入二维数组和表格处理。

## 1. `if`：闰年判断

通过逻辑运算组合规则。

## 2. 多分支判断

理解 `if...elif...else` 从上到下匹配。


```python
#if elif else的区别
if a>b:
    print("a>b")
elif a==b:
    print("a=b")
else:
    print("a<b")
```

## 3. 嵌套 `if`

先验证三角形是否合法，再继续分类。


```python
'''if的高阶使用
#输入三角形的三边，判断是等边三角形，还是直角三角形还是普通三角形，或者不能构成三角形
# 获取用户输入的三个整数，分别代表三角形的三条边长
'''
a,b,c=map(int,input("请输入三角形的三边：").split())
# 判断三条边是否满足三角形不等式定理（任意两边之和大于第三边）
if a+b>c and a+c>b and b+c>a:
    # 判断是否为等边三角形（三边相等）
    if a==b and b==c:
        print("等边三角形")
    # 判断是否为直角三角形（满足勾股定理）
    elif a**2+b**2==c**2 or a**2+c**2==b**2 or b**2+c**2==a**2:
        print("直角三角形")
    # 既非等边也非直角的普通三角形
    else:
        print("普通三角形")
# 不满足三角形构成条件
else:
    print("不能构成三角形")
```

## 4. `match...case`

适合对一个值匹配多个明确情况。


```python
#match...case 语句
match_str = input("请输入一个字符串:")
match match_str:
    case "hello world":
        print("hello world")
    case "hello":
        print("hello")
    case "world":
        print("world")
    case _:
        print("other")
```

## 5. `while` 循环

重点理解初始化、条件、循环体、状态更新。


```python
#循环语句while
a=10
while a>0:#判断a是否大于0，大于0则执行循环体，小于0则结束循环
    print(a)
    a-=1
```

## 6. `while` 累加

使用循环变量与累加器保存运行状态。


```python
#计算1-100之间的偶数和
sum=0
i=1
while i<=100:
    if i %2==0:
        sum+=i
    i+=1
print(sum)
```

## 7. `for` 与 `range()`

`range()` 生成可迭代范围，结束值不包含在内。


```python
#for循环本质是轮询机制
for i in range(2,20):#生成2-19的数字
    print(i)

```

## 8. 条件筛选与累加

在遍历中加入条件，形成“遍历 → 判断 → 聚合”的模式。


```python
# 计算1-100所有的三的倍数之和
# 初始化总和为0
sum = 0
# 遍历1到100的整数，累加所有能被3整除的数
for i in range(1, 101):
    if i % 3 == 0:
        sum += i
# 打印最终的累加结果
print(sum)
```

## 9. 嵌套循环

外层控制行，内层控制列。


```python
#打印一个长度为10.宽度为5的长方形
m=input("请输入长方形的长度")
n=input("请输入长方形的宽度")
for i in range(int(m)):
    for j in range(int(n)):
        print("*",end=" ")
    print("\n")
    
```

## 10. `break` 与 `continue`

补齐循环流程控制。


```python
for number in range(1, 10):
    # 偶数跳过当前轮
    if number % 2 == 0:
        continue

    # 大于7后结束循环
    if number > 7:
        break

    print(number)
```

## 11. 无限循环退出设计

`while True` 本身不是错误，关键是必须有明确退出条件。


```python
count = 0

while True:
    count += 1
    print(count)

    # 达到3次后主动结束
    if count >= 3:
        break

print("循环结束")
```

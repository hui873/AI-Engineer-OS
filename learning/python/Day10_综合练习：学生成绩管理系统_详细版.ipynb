# 学习内容与代码实践


## 1. 基础版学生成绩管理系统

仅使用当前已学基础内容，不提前引入文件、类、测试等后续工程知识。


```python
students = []

while True:
    print("\n=== 学生成绩管理系统 ===")
    print("1. 新增学生")
    print("2. 查询学生")
    print("3. 显示全部学生")
    print("4. 退出")

    choice = input("请选择功能：").strip()

    if choice == "1":
        student_id = input("请输入学号：").strip()
        name = input("请输入姓名：").strip()
        score_text = input("请输入成绩：").strip()

        # 检查学号是否重复
        duplicate = False
        for student in students:
            if student["student_id"] == student_id:
                duplicate = True
                break

        if duplicate:
            print("学号已存在")
            continue

        # 基础输入校验
        if not score_text.isdigit():
            print("成绩必须是整数")
            continue

        score = int(score_text)

        if score < 0 or score > 100:
            print("成绩必须在0到100之间")
            continue

        student = {
            "student_id": student_id,
            "name": name,
            "score": score
        }

        students.append(student)
        print("新增成功")

    elif choice == "2":
        student_id = input("请输入要查询的学号：").strip()

        found = None

        for student in students:
            if student["student_id"] == student_id:
                found = student
                break

        if found:
            print(found)
        else:
            print("未找到该学生")

    elif choice == "3":
        if not students:
            print("暂无学生数据")
            continue

        total_score = 0

        for student in students:
            score = student["score"]
            total_score += score

            if score >= 90:
                level = "优秀"
            elif score >= 80:
                level = "良好"
            elif score >= 60:
                level = "及格"
            else:
                level = "不及格"

            print(
                student["student_id"],
                student["name"],
                student["score"],
                level
            )

        average = total_score / len(students)
        print(f"班级平均分：{average:.2f}")

    elif choice == "4":
        print("程序结束")
        break

    else:
        print("无效选择，请重新输入")
```

# Day 8.2 - Python文件操作（File I/O）

## 今日总结

今天学习了 Python 工程开发中的文件操作，理解了内存数据、外部文件、资源管理和文件路径之间的关系。

**主要掌握：**

* 文件与内存数据的区别
* 使用 `open()` 打开文件
* 使用 `read()` 读取全部内容
* 使用 `readline()` 逐行读取内容
* 使用 `readlines()` 将多行内容读取为列表
* 使用 `write()` 向文件写入数据
* 文件模式 `"r"`、`"w"`、`"a"` 的作用
* 使用 `with open()` 自动管理文件资源
* 使用 `encoding="utf-8"` 处理中文文本
* 相对路径与绝对路径的区别
* 使用相对路径提高项目的可移植性
* 文件操作与数据持久化的关系

**详细理解：**

1. **文件能够实现数据持久化**

   理解：程序中的变量通常保存在内存中，程序结束后数据会消失，而文件可以将数据保存到硬盘，使程序重新运行后仍然能够读取之前的数据。

2. **文件模式决定数据的操作方式**

   理解：`"r"` 用于读取文件，`"w"` 用于重新写入并覆盖原有内容，`"a"` 用于在文件末尾追加内容，因此需要根据实际的数据处理需求选择合适的模式。

3. **`with open()`能够安全管理文件资源**

   理解：直接使用 `open()` 后需要手动调用 `close()`，而 `with open()` 可以在代码执行结束或发生异常时自动关闭文件，减少资源未释放的问题。

4. **相对路径能够提高项目可移植性**

   理解：绝对路径依赖当前电脑的目录结构，而相对路径根据项目位置定位文件，更适合代码迁移、GitHub 共享和服务器部署。

5. **文件编码影响文本数据的正确读取**

   理解：读取和写入中文文件时，需要统一使用 `encoding="utf-8"`，否则可能出现乱码或编码异常，因此编码也是文件操作中需要明确管理的一部分。

**今天最大的收获：**

> 文件操作不只是读取和写入内容，还需要同时考虑数据持久化、资源管理、编码方式和项目路径的可移植性。

**通过今天的学习，我理解了：**

* 内存中的数据是临时状态，文件中的数据可以长期保存
* `with open()` 不只是简化语法，而是在帮助程序安全管理文件资源
* 不同文件模式会直接影响原有数据是否被覆盖
* 相对路径能够减少程序对个人电脑环境的依赖
* 文件操作是后续学习 JSON、配置文件和数据集读取的基础

**今天开始从：**

> 只关注文件内容能不能被成功读取和写入

**转向：**

> 开始关注文件资源、数据安全、编码方式以及代码在不同运行环境中的可用性。


## 2. 文件写入与读取

基础读写。


```python
with open("patient.txt", "w", encoding="utf-8") as file:
    file.write("P001,Tom,30\n")

with open("patient.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## 3. 逐行读取

避免大文件一次全部载入内存。


```python
with open("patient.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())
```

## 4. 追加模式

在原文件末尾增加内容。


```python
with open("patient.txt", "a", encoding="utf-8") as file:
    file.write("P002,Alice,25\n")
```

## 5. 自定义上下文管理器

理解 `with` 背后的协议。


```python
class Resource:
    def __enter__(self):
        print("获取资源")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("释放资源")

with Resource() as resource:
    print("使用资源")
```

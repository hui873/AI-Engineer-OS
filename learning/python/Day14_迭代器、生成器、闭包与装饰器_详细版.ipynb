# 学习内容与代码实践


## 1. `iter()` 与 `next()`

手动观察for循环背后的迭代过程。


```python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))

# 再调用一次会触发StopIteration
# print(next(iterator))
```

## 2. 自定义迭代器

实现 `__iter__()` 和 `__next__()`。


```python
class Counter:
    def __init__(self, limit):
        self.current = 0
        self.limit = limit

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= self.limit:
            raise StopIteration

        value = self.current
        self.current += 1
        return value

for number in Counter(3):
    print(number)
```

## 3. 生成器与 `yield`

按需产生数据。


```python
def patient_ids(count):
    for index in range(1, count + 1):
        yield f"P{index:03d}"

generator = patient_ids(3)

print(next(generator))
print(next(generator))

for patient_id in generator:
    print(patient_id)
```

## 4. 闭包

内部函数保留外层变量。


```python
def create_multiplier(factor):
    def multiply(value):
        return value * factor

    return multiply

double = create_multiplier(2)
triple = create_multiplier(3)

print(double(10))
print(triple(10))
```

## 5. 基础装饰器

在不修改原函数内部逻辑的情况下增加行为。


```python
from functools import wraps

def log_call(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"开始调用：{func.__name__}")

        result = func(*args, **kwargs)

        print(f"调用结束：{func.__name__}")
        return result

    return wrapper

@log_call
def calculate_bmi(weight, height):
    return weight / (height ** 2)

print(calculate_bmi(70, 1.75))
```

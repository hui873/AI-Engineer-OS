3. **set元素必须可哈希**

   理解：与dict key相同，这是哈希查找结构的要求。

4. **frozenset提供不可变集合**

   理解：当集合本身需要被哈希或作为key时非常有用。

## 2. 去重与成员判断

最常用场景。


```python
patient_ids = ["P001", "P002", "P001", "P003"]

unique_ids = set(patient_ids)

print(unique_ids)
print("P002" in unique_ids)
```

## 3. 集合运算

并集、交集、差集、对称差集。


```python
a = {"P001", "P002", "P003"}
b = {"P002", "P003", "P004"}

print(a | b)
print(a & b)
print(a - b)
print(a ^ b)
```

## 4. 子集与超集

判断包含关系。


```python
critical = {"P001", "P002"}
all_patients = {"P001", "P002", "P003"}

print(critical.issubset(all_patients))
print(all_patients.issuperset(critical))
```

## 5. `frozenset`

不可修改的集合。


```python
required_fields = frozenset({
    "name",
    "age",
    "disease"
})

print(required_fields)

# required_fields.add("phone")
# frozenset没有add方法
```

## 1. 类、对象与初始化

保留原Car类代码。


```python
# 面向对象编程
# 对象可以理解为现实中具体的人/物在程序中的数字化身（万物皆对象）。
# 它把一个人/物的特征(属性)和功能(方法)打包到一起，是面向对象编程的基本单元。（关注的是谁来帮我做这件事儿）
# 类：描述的是一组具有相同属性（特征）和方法（功能/行为）的模板
# 对象：对象是类的实例，是基于类创建出来的（实例对象）
# 说明：类名的命名规范，遵循大驼峰命名法，每个单词的首字母都是大写，单词之间没有分隔符，比如：UserInfo，UserAccount。
# 对象名=类名()）
# 对象名.属性名1=属性值1
# 对象名.属性名2=属性值2
class Car:
    def __init__(self, c_brand, c_name, c_price):#__init__：初始化方法，对象创建后自动调用，主要用于设置对象的初始状态（设置对象属性）
        """
        初始化方法，对象创建后自动调用，主要用于设置对象的初始状态（设置对象属性）
        :param c_brand: 车的品牌
        :param c_name: 车的名称
        :param c_price: 车的价格
        """
        self.brand = c_brand
        self.name = c_name #self：方法的第一个参数，表示当前创建的实例对象
        self.price= c_price
#创建对象
c1 = Car(c_brand="BMW", c_name="X5", c_price=500000)
print(c1.__dict__)
```

## 3. 属性查找

补充实例与类属性同名时的基本行为。


```python
class Patient:
    status = "normal"

patient = Patient()

# 实例中没有status时，会读取类属性
print(patient.status)

# 创建同名实例属性后，会优先读取实例自己的值
patient.status = "critical"

print(patient.status)
print(Patient.status)
```

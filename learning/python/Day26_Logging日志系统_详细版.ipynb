# 学习内容与代码实践


## 1. 日志等级与logger

模块级logger。


```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s | %(levelname)s | %(name)s | %(message)s"
)

logger = logging.getLogger(__name__)

logger.debug("调试信息")
logger.info("程序启动")
logger.warning("检测到需要关注的数据")
logger.error("当前操作执行失败")
```

## 2. FileHandler

把日志写入文件。


```python
import logging

logger = logging.getLogger("medical_app")
logger.setLevel(logging.INFO)

file_handler = logging.FileHandler(
    "app.log",
    encoding="utf-8"
)

formatter = logging.Formatter(
    "%(asctime)s | %(levelname)s | %(name)s | %(message)s"
)

file_handler.setFormatter(formatter)

logger.addHandler(file_handler)

logger.info("患者数据处理开始")
```

## 3. 日志内容边界

哪些信息不应该记录。


### 医疗项目日志特别注意

不应直接记录：

* 密码
* API Key / Token
* 完整身份证号
* 完整手机号
* 不必要的患者隐私字段
* 原始敏感病历全文

日志应该记录足够定位问题的信息，但不能为了调试把敏感数据全部暴露出来。

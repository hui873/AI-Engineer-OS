## 1. 基础异常捕获

保留原代码和注释。


```python
# 异常（也称为Bug）就是程序运行过程中出现的错误，它会中断程序的正常执行流程。
# 作用：保证数据、逻辑的正确性，避免程序执行混乱
# 在开发阶段，尽量发现更多的问题，尽早解决问题，保障程序正常执行

try:
    a = 10 / 0
    print(a)
except ZeroDivisionError as h:# 捕获除数为0的异常，打印异常信息h
    print("除数不能为0",h)
except Exception as e:# 捕获其他异常，打印异常信息e
    print("发生异常:", e)
finally:
    print("程序执行结束")
# 程序运行过程中出现异常，有两种处理方案：
# 1．不做处理：整个程序因为一个Bug，中断执行。
# 2，捕获异常：按照我们自己的处理方式，处理完异常，程序继续执行。
```

## 3. 自定义异常

保留原MyError代码。


```python
#自定义异常的例子
class MyError(Exception):
    """自定义异常类"""

    def __init__(self, message, code=1001):
        """
        初始化自定义异常对象。

        :param message: 异常描述信息
        :param code: 异常错误码，默认为 1001
        """
        self.message = message  # 保存异常说明
        self.code = code        # 保存异常错误码
        super().__init__(message)  # 调用父类 Exception 的初始化方法


def check_age(age):
    """
    检查年龄是否合法。

    :param age: 输入的年龄
    :return: 如果年龄合法，返回提示信息
    :raises MyError: 当年龄不合法时抛出自定义异常
    """
    if age < 0:
        # 如果年龄小于 0，说明输入不合法
        raise MyError("年龄不能为负数", code=1001)

    if age < 18:
        # 如果年龄小于 18，说明用户未成年
        raise MyError("未成年人不能使用", code=1002)

    # 以上条件都不满足，说明年龄合法
    return "允许使用"


try:
    # 尝试调用函数，并传入一个非法年龄
    print(check_age(-3))
except MyError as e:
    # 捕获自定义异常，并输出异常信息
    print(f"捕获到异常：{e.message}")
    print(f"错误码：{e.code}")
```

## 4. `else` 和常见异常

补齐执行流程。


```python
try:
    age = int("30")
except ValueError:
    print("年龄格式错误")
else:
    print(f"年龄转换成功：{age}")
finally:
    print("本次处理结束")

try:
    int([1, 2, 3])
except TypeError as e:
    print(type(e).__name__, e)
```

## 5. 不要吞掉异常

至少处理、记录或重新抛出。


```python
def parse_age(text):
    try:
        return int(text)
    except ValueError:
        print(f"年龄解析失败：{text}")
        raise

try:
    parse_age("abc")
except ValueError:
    print("由上层决定如何提示用户")
```

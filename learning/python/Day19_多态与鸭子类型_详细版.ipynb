## 1. 继承体系多态

保留原多态代码。


```python
#多态，一个方法，具有不同的表现形式
class Car:
    def __init__(self, c_color, c_brand, c_model):
        self.color = c_color
        self.brand = c_brand
        self.model = c_model

    def start(self):  # 启动
        print(f'{self.brand}{self.model}正在启动...')
    def stop(self):  # 停止
        print(f'{self.brand}{self.model}停止行驶...')
    def run(self):  # 行驶
        print(f'{self.brand}{self.model}正在行驶')
    def charge(self):  # 充电
        print(f'{self.brand}{self.model}正在加燃料')

# 电动车类继承汽车类
class ElectricCar(Car):  
    def charge(self):
        #super().charge()  # 调用父类的charge方法
        print(f'{self.brand}{self.model}正在充电')

class GasCar(Car):
    def charge(self):
        #Car.charge(self)  # 调用父类的charge方法
        print(f'{self.brand}{self.model}正在加汽油')

def  handle_charge(car:Car):
    car.charge()    

if __name__ == "__main__":
    handle_charge(GasCar(c_brand="BMw", c_model="x5", c_color="黑色"))
    handle_charge(ElectricCar(c_brand="BMw", c_model="x5", c_color="黑色"))
```

## 3. 抽象基类

补齐显式接口约束。


```python
from abc import ABC, abstractmethod

class Model(ABC):
    @abstractmethod
    def predict(self, data):
        pass

class RiskModel(Model):
    def predict(self, data):
        return "high" if data >= 80 else "normal"

model = RiskModel()
print(model.predict(90))
```

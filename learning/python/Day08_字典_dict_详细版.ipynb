## 2. `[]` 与 `get()`

理解必须字段与可选字段。


```python
patient = {"name": "Tom", "age": 30}

print(patient["name"])
print(patient.get("phone"))
print(patient.get("phone", "暂无联系方式"))

print("age" in patient)
```

## 3. `update()`

一次修改或添加多个字段。


```python
patient = {"name": "Tom", "age": 30}

patient.update({
    "age": 31,
    "status": "stable",
    "department": "Respiratory"
})

print(patient)
```

## 4. `setdefault()`

适合初始化不存在的字段。


```python
patient = {"name": "Tom"}

symptoms = patient.setdefault("symptoms", [])
symptoms.append("cough")

print(patient)
```

## 5. 嵌套字典

真实项目经常出现多层结构。


```python
patient = {
    "id": "P001",
    "name": "Tom",
    "vital_signs": {
        "temperature": 37.8,
        "heart_rate": 82
    }
}

print(patient["vital_signs"]["temperature"])
```

## 6. 字典推导式

用于简单规则下快速生成映射。


```python
names = ["Tom", "Alice", "Jerry"]

name_lengths = {
    name: len(name)
    for name in names
}

print(name_lengths)
```

# 学习内容与代码实践


## 1. `match()` 与 `search()`

先理解从开头匹配和全局搜索的区别。


```python
import re

text = "Patient ID: P001"

# match() 从字符串开头尝试匹配
result1 = re.match(r"Patient", text)
print(result1.group() if result1 else None)

# search() 在整个字符串中搜索第一个匹配
result2 = re.search(r"P\d+", text)
print(result2.group() if result2 else None)
```

## 2. `findall()` 批量提取

适合从一段文本中提取所有符合模式的数据。


```python
import re

text = "P001 temperature=36.8, P002 temperature=38.2"

# \d+ 匹配一个或多个数字
patient_ids = re.findall(r"P\d+", text)

# \d+\.\d+ 匹配带小数点的数字
temperatures = re.findall(r"\d+\.\d+", text)

print(patient_ids)
print(temperatures)
```

## 3. `sub()` 文本替换

按模式替换敏感信息或清洗文本。


```python
import re

text = "联系电话：13812345678"

# 将手机号中间四位替换为****
masked = re.sub(
    r"(\d{3})\d{4}(\d{4})",
    r"\1****\2",
    text
)

print(masked)
```

## 4. 普通字符串邮箱校验

复习之前的字符串规则思路。


```python
email = "user@example.com"

is_basic_valid = (
    email.count("@") == 1
    and "." in email.split("@")[-1]
    and not email.startswith(("@", "."))
    and not email.endswith(("@", "."))
)

print(is_basic_valid)
```

## 5. 正则邮箱校验对比

这里只做基础示例，目的是理解表达方式差异，不代表完整RFC邮箱标准。


```python
import re

email = "user@example.com"

pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

if re.match(pattern, email):
    print("邮箱格式基本合法")
else:
    print("邮箱格式有问题")
```

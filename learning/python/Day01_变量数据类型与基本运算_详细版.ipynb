**详细理解：**

1. **变量是名称与对象之间的绑定**

   理解：变量不是固定类型的盒子，而是一个名称当前指向某个对象。重新赋值时，变量名会重新绑定到新的对象。

2. **数据类型决定运算方式**

   理解：字符串 `"18"` 和整数 `18` 虽然看起来相似，但它们参与运算时规则完全不同，因此外部数据进入程序后要先明确类型。

3. **`input()`得到的始终是字符串**

   理解：用户输入年龄、金额等数字时，程序仍然会先得到字符串，因此参与数值计算前通常需要 `int()` 或 `float()` 转换。

4. **运算符共同组成程序规则**

   理解：算术运算用于计算，比较运算产生布尔值，逻辑运算再组合多个条件，为后面的流程控制打基础。

5. **浮点数存在精度边界**

   理解：计算机使用二进制表示浮点数，因此部分十进制小数无法被完全精确表示，实际判断时有时需要使用近似比较。

# 学习内容与代码实践


## 1. 变量与重新赋值

变量保存程序状态，重新赋值表示变量名重新绑定新的值。


```python
# 初始化浮点数变量num并赋值为114.1
num=114.1
print(num)
# 更新num的值为114.5
num = 114.5
print(num)
```

## 2. 布尔类型

`bool` 只有 `True` 与 `False` 两个值，是条件判断的核心类型。


```python
a=True
print(a)
```

## 3. 多变量赋值与变量交换

Python 可以一次为多个变量赋值，也可以直接交换变量。


```python
a=10
b=20
a,b=b,a
print(a,b)
```

## 4. f-string

f-string 可以把变量直接嵌入字符串，是 Python 中非常常用的格式化方式。


```python
name = 'zhanghui'
age = '18'
print(f"I'm {name},jinnian {age}")
```

## 5. 输入与输出

`input()` 接收用户输入，`print()` 输出结果。注意 `input()` 返回值始终是字符串。


```python
#输入 input()函数，输出print(),input()获得的都是str类型
a='zhanghui'
print(f'我是{a}')
```

## 6. 输入数据的类型转换

当输入内容需要参与数值计算时，应先转换成正确类型。


```python
Yue=10000
mima=input("请输入你的密码")
quqian=input("请输入取款金额")
print(Yue-int(quqian))
```

## 7. 算术与复合赋值运算

掌握 `+ - * / // % **`，以及 `+=` 等复合赋值。


```python
#运算符，主要是/是小数，//是整数，%是取余，出现float运算时，可能会出现精度损失
#num=num+1等效于num+=1
a=2
a+=1
print(f'a的值变为了{a}')
```

## 8. 比较与逻辑运算

比较表达式会产生布尔值，`and/or/not` 可以组合多个条件。


```python
#逻辑运算符 and，or,not 即与或非
a=int(input("请输入"))
if a>10 and a<20:
#if 10<a<20 等价写法
    print(f'{a}在10到20之间')
else:
    print(f'{a}不在10到20之间')
```

## 9. `type()` 与动态类型

补充 `type()`，帮助理解变量当前绑定对象的类型。


```python
age = 18
height = 1.75
name = "Tom"
is_patient = True

# 查看不同变量当前值的数据类型
print(type(age))
print(type(height))
print(type(name))
print(type(is_patient))

# Python变量可以重新绑定不同类型的对象
value = 100
print(type(value))

value = "100"
print(type(value))
```

## 10. 浮点数精度

浮点数是近似表示，不能默认所有小数运算都完全精确。


```python
import math

result = 0.1 + 0.2

# 结果通常不是精确的0.3
print(result)

# 对浮点结果做近似比较
print(math.isclose(result, 0.3))
```

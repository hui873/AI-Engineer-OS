## 1. 九九乘法表

通过嵌套循环生成二维规律。


```python
#打印九九乘法表
# 外层循环：控制乘数，从1到9
for i in range(1, 10):
    # 内层循环：控制被乘数，从1到当前的i值
    for j in range(1, i + 1):
        # 打印乘法表达式，使用制表符\t分隔，使输出对齐
        print(f"{i}*{j}={i*j}", end="\t")
    # 内层循环结束后，打印空行换行，开始下一行
    print()
```

## 2. 登录系统

通过 `while`、条件和 `break/continue` 实现交互。


```python
#根据用户输入执行登录操作
# 使用while 1创建无限循环，直到登录成功
while 1:
    # 获取用户输入的用户名
    a = input("请输入用户名：")
    # 获取用户输入的密码
    b = input("请输入密码：")
    # 判断用户名和密码是否匹配
    # 有三组有效的用户名和密码组合：
    # 1. admin / 666888
    # 2. zhanasan / 123456
    # 3. user / 888666
    if (a == "admin" and b == "666888") or (a == "zhanasan" and b == "123456") or (a == "user" and b == "888666"):
        # 用户名和密码匹配，显示登录成功
        print("登录成功！")
        # 跳出循环，结束程序
        break
    # 判断用户名或密码是否为空
    elif a == "" or b == "":
        # 提示用户名或密码不能为空
        print("用户名或密码不能为空！")
        # 跳过本次循环剩余代码，直接进入下一次循环
        continue
    # 用户名和密码不匹配且不为空
    else:
        # 显示登录失败提示
        print("登录失败！")


        
```

## 3. 猜数字游戏

结合 `random` 和循环反馈。


```python
import random
#猜数字小游戏
# 生成一个1到100之间的随机整数
a = random.randint(1, 100)
# 使用while True创建无限循环，直到猜对为止
while True:
    # 获取用户输入的数字，并转换为整数类型
    b = int(input("请输入数字："))
    # 判断用户输入的数字是否大于目标数字
    if b > a:
        # 如果大于，提示太大了
        print("太大了")
    # 判断用户输入的数字是否小于目标数字
    elif b < a:
        # 如果小于，提示太小了
        print("太小了")
    # 如果既不大于也不小于，说明猜对了
    else:
        # 祝贺用户猜对了
        print("恭喜你猜对了")
        # 显示正确答案
        print(a)
        # 跳出循环，结束游戏
        break
```

## 4. 输入校验

补充用户输入不是整数时的处理。


```python
while True:
    value = input("请输入整数，输入q退出：")

    if value == "q":
        break

    try:
        number = int(value)
        print(f"你输入的是：{number}")
    except ValueError:
        print("输入无效，请输入整数")
```

## 5. 程序状态与尝试次数

用计数器显式管理程序运行状态。


```python
max_attempts = 3

for attempt in range(1, max_attempts + 1):
    print(f"当前是第{attempt}次尝试")

    if attempt == max_attempts:
        print("达到最大尝试次数")
```

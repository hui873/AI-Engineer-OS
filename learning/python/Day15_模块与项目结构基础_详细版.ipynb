## 4. 循环依赖示意

理解问题，不需要为了演示真的创建错误模块。


```text
不推荐：

a.py  -> import b
b.py  -> import a
```

如果 A 和 B 都依赖同一段逻辑，可以考虑把共同部分提取到 `common.py`，让依赖方向变成：

```text
a.py ----          -> common.py
b.py ----/
```

核心不是“记住如何绕过循环导入”，而是让模块职责更清晰。

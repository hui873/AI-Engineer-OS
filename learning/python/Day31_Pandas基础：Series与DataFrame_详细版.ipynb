## 1. Series基础

保留原代码。


```python
import pandas as pd
series = pd.Series([1, 2, 3])#创建一个Series对象，一个series就是一列数据，是一维的
print(series)
series = pd.Series([1, 2, 3], index=['a', 'b', 'c'])#创建一个Series对象,索引为a,b,c
print(series)
series = pd.Series([1, 2, 3], index=['a', 'b', 'c'], name='series')
#创建一个Series对象,索引为a,b,c,名称为series
print(series)
print(series.loc['a']) #通过索引获取元素
print(series.iloc[0]) #通过位置获取元素,等价于series.loc['a']
print(series[series > 1]) #通过布尔索引获取元素
```

## 2. 字典创建Series

保留原代码。


```python
import pandas as pd
calories = {"Day1": 1750,"Day 2":2100,"Day3": 1700}#创建一个字典,键为天数,值为卡路里
series = pd.Series(calories)#创建一个Series对象,索引为天数,值为卡路里
print(series)
series.loc['Day1']+=500
print(series)
```

## 3. DataFrame基础

保留原代码。


```python
#dataframe是一个二维的表格数据结构,类似于Excel表格,可以看作是由多个Series组成的字典
import pandas as pd
data={
"Name":["Spongebob", "Patrick", "Squidward"],
"Age":[30,35,50]
}
df = pd.DataFrame(data)
print(df)
df = pd.DataFrame(data, index=['a', 'b', 'c'])#创建一个DataFrame对象,索引为a,b,c
print(df)
print(df.loc['a']) #通过索引获取行
print(df.iloc[0]) #通过位置获取行,等价于df.loc['a']
df["Job"] = ["Cook","N/A","Cashier"]#添加一列数据,索引为a,b,c
print(df)
new_row = pd.DataFrame([{"Name": "Sandy", "Age": 28, "Job": "Engineer"}], index=['d'])
df = pd.concat([df, new_row])#添加一行数据,索引为d
print(df)
#可以使用pd.read_csv()和pd.read_json()函数读取csv和json文件 
#df = pd.read_csv('data.csv')
```

## 4. CSV与loc/iloc

保留原代码。


```python
import pandas as pd
#读入csv文件
df = pd.read_csv('data.csv',index_col='Name') #将Name列作为索引
#print(df)
#print(df['Name'])#获取Name列数据
#print(df[['Height','Name']])#获取Height和Name列数据
print(df.loc["Pikachu"])#获取Name为Pikachu的行数据
#获取Name为Moltres到Mewtwo的行数据,并只获取Height和Weight列数据
print(df.loc["Moltres":"Mewtwo",["Height","Weight"]])
print(df.iloc[0:11:2,0:3])#获取索引为0,2,4,6,8,10的行数据,并只获取0-2列数据
```

# Day 8.3 - JSON数据处理（JSON）

## 今日总结

今天学习了 Python 工程开发中的 JSON 数据处理，理解了程序对象、基础数据结构与外部文件之间的转换关系。

**主要掌握：**

* JSON 的基本概念和工程作用
* Python 数据类型与 JSON 数据类型的对应关系
* 使用 `json.dump()` 将数据保存到 JSON 文件
* 使用 `json.load()` 从 JSON 文件读取数据
* 将列表数据保存为 JSON
* 自定义对象不能直接序列化为 JSON
* 使用 `to_dict()` 将对象转换为字典
* 使用 `from_dict()` 将字典恢复为对象
* 使用 `indent=4` 格式化 JSON 文件
* 使用 `ensure_ascii=False` 正常显示中文

**详细理解：**

1. **JSON作为数据交换和存储格式**

   理解：JSON 可以把程序中的字典、列表等数据保存到外部文件，也可以在不同程序之间传递数据，是程序内部数据与外部存储之间的重要桥梁。

2. **Python数据与JSON数据需要进行转换**

   理解：Python 中的字典、列表、字符串、数字、布尔值和 `None` 可以转换为 JSON 支持的数据类型，但两者属于不同的数据表示形式，需要通过序列化和反序列化完成转换。

3. **`json.dump()`与`json.load()`负责文件读写**

   理解：`json.dump()` 负责把 Python 数据转换并写入 JSON 文件，`json.load()` 负责读取 JSON 文件并恢复为 Python 数据，使程序能够保存和重新加载运行状态。

4. **自定义对象不能直接保存为JSON**

   理解：JSON 无法识别 `Patient` 等自定义类的对象，因此需要先通过 `to_dict()` 转换为字典，读取后再通过 `from_dict()` 恢复为对象。

5. **JSON文件也需要考虑可读性和中文显示**

   理解：使用 `indent=4` 可以让 JSON 数据结构更加清晰，使用 `ensure_ascii=False` 可以让中文直接显示，从而提高数据文件的可读性和可维护性。

**今天最大的收获：**

> 数据保存不是简单地把对象写入文件，而是需要设计对象、基础数据结构与存储格式之间的转换关系。

**通过今天的学习，我理解了：**

* 程序中的对象和文件中的数据属于不同的数据表示形式
* 序列化负责把程序数据转换为可存储的格式
* 反序列化负责把外部数据恢复为程序可以使用的结构
* `to_dict()` 和 `from_dict()` 是对象模型与数据存储之间的桥梁
* 数据文件不仅要能够正确读写，还要考虑可读性和后续维护

**今天开始从：**

> 认为程序中的对象可以直接保存到文件

**转向：**

> 理解对象需要先转换为 JSON 支持的基础数据结构，读取后再重新构建为程序对象。


## 2. `dump()` 与 `load()`

文件级序列化。


```python
import json

patient = {
    "name": "Tom",
    "age": 30,
    "symptoms": ["cough", "fever"]
}

with open("patient.json", "w", encoding="utf-8") as file:
    json.dump(
        patient,
        file,
        ensure_ascii=False,
        indent=4
    )

with open("patient.json", "r", encoding="utf-8") as file:
    loaded = json.load(file)

print(loaded)
```

## 3. `dumps()` 与 `loads()`

字符串级序列化。


```python
import json

data = {"name": "Tom", "age": 30}

text = json.dumps(data, ensure_ascii=False)
print(text)

restored = json.loads(text)
print(restored)
```

## 4. 自定义对象转换

使用 `to_dict()` 与 `from_dict()`。


```python
class Patient:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def to_dict(self):
        return {
            "name": self.name,
            "age": self.age
        }

    @classmethod
    def from_dict(cls, data):
        return cls(
            name=data["name"],
            age=data["age"]
        )

patient = Patient("Tom", 30)

data = patient.to_dict()
restored = Patient.from_dict(data)

print(restored.name, restored.age)
```

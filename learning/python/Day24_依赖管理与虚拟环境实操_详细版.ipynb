**详细理解：**

1. **虚拟环境隔离项目依赖**

   理解：不同项目可以使用不同版本的库。

2. **依赖文件帮助环境复现**

   理解：别人拿到项目后应能安装相同依赖。

3. **配置不应全部硬编码**

   理解：API Key、密码和环境差异应从外部配置读取。

4. **.env不是天然安全**

   理解：安全来自不提交敏感信息并正确管理权限。

## 1. 虚拟环境与依赖常用命令

建议在终端真实操作。


### 常用命令

* `python -m venv venv`
* Windows：`venv\Scripts\activate`
* `pip install package`
* `pip list`
* `pip freeze > requirements.txt`
* `pip install -r requirements.txt`

建议每个项目使用自己的虚拟环境。

## 2. `.gitignore` 基本内容

避免把环境、缓存和敏感配置提交仓库。


常见忽略项：

```text
venv/
.venv/
__pycache__/
*.pyc
.env
*.log
```

`.gitignore` 只是告诉 Git 不追踪这些文件，并不是安全机制。

## 3. 类方法与静态方法

补齐必须掌握的实际用法。


```python
class Patient:
    hospital = "AI Hospital"

    def __init__(self, name):
        self.name = name

    @classmethod
    def change_hospital(cls, hospital):
        cls.hospital = hospital

    @staticmethod
    def is_valid_age(age):
        return 0 <= age <= 120

Patient.change_hospital("Medical Center")

print(Patient.hospital)
print(Patient.is_valid_age(30))
```

## 4. `property`

在属性访问时加入验证。


```python
class Patient:
    def __init__(self, age):
        self.age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("年龄不能为负数")

        self._age = value

patient = Patient(30)

print(patient.age)

# patient.age = -1
# 会触发ValueError
```

## 5. `__str__` 与 `__call__`

让对象具备更自然的Python行为。


```python
class RiskModel:
    def __str__(self):
        return "RiskModel(v1)"

    def __call__(self, score):
        return "high" if score >= 80 else "normal"

model = RiskModel()

print(model)
print(model(90))
```

# 学习内容与代码实践


## 1. 标量运算

保留原代码。


```python
import numpy as np
#numpy的标量计算
array = np.array([1, 2, 3])
print(array + 1) #每个元素加1
print(array * 2) #每个元素乘以2
print(array **3) #每个元素的立方
```

## 2. 向量运算

保留原代码。


```python
import numpy as np
#numpy的向量计算
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
print(array1 + array2) #对应元素相加    
print(array1 * array2) #对应元素相乘
print(np.pi*array1**2) #每个元素的圆的面积
print(np.sqrt(array1)) #每个元素的平方根
array1[array1>2] = 0 #将大于2的元素赋值为0
print(array1)
```

## 3. 广播

保留原代码。


```python
import numpy as np
#numpy的广播机制,从右往左匹配相同或者其中一个数组有维度为1，则可以广播
array1 = np.array([[1, 2, 3], [4, 5, 6]])
array2 = np.array([1, 2, 3])
print(array1.shape) #查看数组的形状
print(array2.shape) #查看数组的形状
print(array1 + array2) #对应元素相加
print(array1 * array2) #对应元素相乘
```

## 4. 聚合

保留原代码。


```python
import numpy as np
#常用的聚合函数
array = np.array([[1, 2, 3], [4, 5, 6]])
print(np.sum(array)) #求和
print(np.sum(array, axis=0)) #求列方向和
print(np.mean(array)) #求平均值
```

## 5. 条件筛选

原代码和注释保持不变。注意：NumPy中使用 `&` 不是因为“底层是C”，而是比较结果是布尔数组，`&` 被重载为逐元素组合；Python `and` 要求整体真值。


```python
import numpy as np
#过滤数组
ages=np.array([[21,17,19,20,16,30,18,65],[39,22,15,99,18,19,20,21]])
teenagers = ages[ages < 18]
print(teenagers)
adults = ages[(ages >= 18) & (ages < 65)]#numpy是基于C语言实现的，所以要用&不能用and
print(adults)
#置换
print(np.where(ages < 18, 0, ages)) #将小于18的元素置为0
```

## 6. 保存与读取

原代码和注释保持不变。注意：`np.array('[1, 2, 3], [4, 5, 6]')` 创建的是字符串标量数组，不会自动变成二维数值数组。


```python
import numpy as np
#numpy数组的保存和读取
array = np.array([[1, 2, 3], [4, 5, 6]])
np.save('array.md', array) #保存数组    
array2 = np.load('array.md.npy') #加载数组，numpy会自动在文件名后加上.npy
print(array2)
array3 = np.array('[1, 2, 3], [4, 5, 6]') 
np.savez('array.npz', array2, array3) #保存多个数组
array4 = np.load('array.npz') #加载多个数组
print(array4['arr_0']) #加载第一个数组
print(array4['arr_1']) #加载第二个数组
```

## 7. `axis=0` 与 `axis=1`

补齐轴概念。


```python
import numpy as np

array = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# 压缩第0轴：每列聚合
print(np.sum(array, axis=0))

# 压缩第1轴：每行聚合
print(np.sum(array, axis=1))
```

## 8. 转置与矩阵乘法

PyTorch前必须掌握。


```python
import numpy as np

a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

print(a.T)

# (2, 3) @ (3, 2) -> (2, 2)
print(a @ b)

# * 是逐元素操作
print(a * 2)
```

## 9. view与copy

理解数据共享。


```python
import numpy as np

original = np.array([1, 2, 3, 4])

# 切片通常是view
view = original[1:3]
view[0] = 99

print(original)
print(view)

# copy创建独立数组
copied = original[1:3].copy()
copied[0] = 1000

print(original)
print(copied)
```

# Day 32 - Python代码规范（PEP8）


## 今日总结

今天学习了 Python 代码规范，理解了为什么工程项目不仅要求代码能够运行，还需要保证代码具有良好的可读性和长期维护能力。


**主要掌握：**

- PEP8代码规范的作用
- Python变量和函数命名规则
- 使用 `小写+下划线` 命名变量和函数
- 使用 `大驼峰命名法` 命名类
- 使用大写字母命名常量
- Python缩进规范
- 函数单一职责原则
- 避免重复代码
- 注释应该解释代码目的而不是代码本身
- 文件和模块命名规范
- Python文件结构组织方式
- 使用代码格式化工具提高代码质量
- 理解代码可读性的重要性


**详细理解：**

1. **代码规范降低团队协作成本**

   理解：在大型项目中，不同开发者需要共同维护代码。如果每个人使用不同的命名和格式，会增加理解成本。统一规范可以让代码更加容易阅读。


2. **命名方式体现代码设计思想**

   理解：变量、函数和类的命名不仅影响阅读体验，也体现代码结构。例如函数应该描述行为，类应该描述对象，让其他开发者快速理解代码作用。


3. **单一职责让代码更容易维护**

   理解：一个函数或类应该只负责一个明确任务。如果一个函数同时处理数据读取、验证和保存，后续修改会更加困难。


4. **代码质量不仅依靠运行结果**

   理解：优秀代码不仅要运行正确，还应该具备清晰结构、合理命名和良好维护性，这也是工程开发与简单编程的重要区别。


5. **自动化工具帮助保持代码规范**

   理解：实际项目中不会完全依靠人工检查代码，而会使用格式化和检查工具保证代码风格统一。


**今天最大的收获：**

> 代码不仅是给计算机执行的指令，也是给开发者阅读和维护的工程文档。


**通过今天的学习，我理解了：**

- 可读性是工程代码的重要质量指标
- 好的命名能够减少代码解释成本
- 代码结构设计会影响项目长期维护
- 自动化工具可以帮助团队保持代码质量
- 编程能力不仅体现在实现功能，也体现在设计高质量代码


**今天开始从：**

> 只追求代码能够完成任务


**转向：**

> 开始关注代码是否清晰、规范以及是否适合长期维护。

## 2. 单一职责

示例展示函数职责拆分。


````python
ADULT_AGE = 18

def is_adult(age: int) -> bool:
    return age >= ADULT_AGE

def format_patient(name: str, age: int) -> str:
    return f"{name} ({age})"

print(is_adult(30))
print(format_patient("Tom", 30))
````

## 4. Git最小工作流

建议在医疗系统项目中真实使用。


### Git 最小工作流

* `git status`：查看修改状态
* `git diff`：查看具体修改
* `git add <file>`：加入暂存区
* `git commit -m "message"`：创建提交
* `git push`：推送远端
* `git pull`：同步远端
* `git branch`：查看分支
* `git switch -c feature-name`：创建并切换分支

建议：一个明确功能或修复，对应一个清晰 commit。
